**COMPARISON**

```python
a,b=3,4
print("a>b",a>b)
print("a<b",a<b)
print("a==b",a==b)
print("a!=b",a!=b)
print("a>=b",a>=b)
print("a<=b",a<=b)
```

**Logical Operators**

(and - or - not)

```python
a,b,c =True,False, True

print("a and b and c",a and b and c)
print("a or b or c",a or b or c)
print("a or b and c",a or b and c)
print("not a",not a)
```
